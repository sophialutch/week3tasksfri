# Graph 2

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
#
```

```python
df = pd.read_csv('kdramas_2023.csv')
```

```python
genres=('Action', 'Adventure','Business', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Family', 'Fantasy','Food', 'Historical', 'Horror', 'Law', 'Life', 'Medical','Melodrama', 'Military', 'Music', 'Mystery', 'Political','Psychological', 'Romance', 'Sci-Fi', 'Sitcom', 'Sports', 'Supernatural', 'Thriller', 'War', 'Youth')
```

```python
data =[]

for i in genres:
    shows = df[df['Genres'].str.contains(i, na=False)]
    for ep in sorted(shows['Episodes'].unique()):
        avgwatchers = shows[shows['Episodes']== ep]['Watchers'].mean()
        data.append({'genres': i,'Episodes':ep,'Watchers':avgwatchers})
```

```python
df = pd.DataFrame(data)

# area chart code skeleton but with these diff names/values
g = sns.FacetGrid(df, col='genres', hue='genres', col_wrap=4, sharey=False)
g = g.map(plt.plot,'Episodes', 'Watchers')
g = g.map(plt.fill_between, 'Episodes','Watchers',alpha=0.2)
g = g.set_titles("{col_name}")

plt.subplots_adjust(top=0.92)
g.fig.suptitle('Watchers amount across episodes by genre')
plt.show()
```
